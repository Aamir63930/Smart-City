```python
from pyspark.sql.functions import (
    col, coalesce, explode, current_timestamp
)
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number

bronze_table = "workspace.smartair.bronze_openaq_measurements"

bronze = spark.table(bronze_table)

# Extract individual measurements and metadata
prepared = (
    bronze
    .select(
        col("extracted_at_utc").alias("extraction_time"),
        coalesce(
            col("payload.search_center_city"),
            col("payload.city_context.city")
        ).alias("search_center_city"),
        coalesce(
            col("payload.actual_location"),
            col("payload.sensor_context.location")
        ).alias("actual_location"),
        coalesce(
            col("payload.sensor_id"),
            col("payload.sensor_context.sensor_id")
        ).alias("sensor_id"),
        explode(col("payload.response.results")).alias("m")
    )
    .select(
        "extraction_time",
        "search_center_city",
        "actual_location",
        "sensor_id",
        col("m.parameter.name").alias("pollutant"),
        col("m.value").cast("double").alias("concentration"),
        col("m.parameter.units").alias("unit"),
        col("m.period.datetimeFrom.utc").cast("timestamp").alias("measurement_start_utc"),
        col("m.period.datetimeTo.utc").cast("timestamp").alias("measurement_end_utc"),
        col("m.coverage.percentComplete").cast("double").alias("data_completeness"),
        col("m.coverage.observedCount").cast("long").alias("observed_count")
    )
)

# Remove incomplete records
prepared = prepared.filter(
    col("sensor_id").isNotNull() &
    col("search_center_city").isNotNull() &
    col("pollutant").isNotNull() &
    col("measurement_start_utc").isNotNull() &
    col("concentration").isNotNull()
)

# Remove duplicate measurements
window_spec = (
    Window.partitionBy(
        "sensor_id",
        "pollutant",
        "measurement_start_utc",
        "measurement_end_utc"
    ).orderBy(col("extraction_time").desc_nulls_last())
)

prepared = (
    prepared
    .withColumn("rn", row_number().over(window_spec))
    .filter(col("rn") == 1)
    .drop("rn", "extraction_time")
    .withColumn("silver_ingested_at", current_timestamp())
)

print("Prepared Silver records:", prepared.count())

display(
    prepared.groupBy("search_center_city")
    .count()
    .orderBy("search_center_city")
)

display(prepared.limit(10))
```

```python
from delta.tables import DeltaTable

silver_table = "workspace.smartair.silver_openaq_measurements"

# Existing Silver count
before_count = spark.table(silver_table).count()

# Prepared data ko temporary view banao
prepared.createOrReplaceTempView("new_silver_measurements")

# Insert only records that do not already exist
spark.sql(f"""
MERGE INTO {silver_table} AS target
USING new_silver_measurements AS source
ON  target.sensor_id = source.sensor_id
AND lower(target.pollutant) = lower(source.pollutant)
AND target.measurement_start_utc = source.measurement_start_utc

WHEN NOT MATCHED THEN INSERT (
    sensor_id,
    actual_location,
    search_center_city,
    pollutant,
    concentration,
    unit,
    measurement_start_utc,
    measurement_end_utc,
    data_completeness,
    observed_count,
    silver_ingested_at
)
VALUES (
    source.sensor_id,
    source.actual_location,
    source.search_center_city,
    source.pollutant,
    source.concentration,
    source.unit,
    source.measurement_start_utc,
    source.measurement_end_utc,
    source.data_completeness,
    source.observed_count,
    source.silver_ingested_at
)
""")

# Verify updated Silver table
after_count = spark.table(silver_table).count()

print("Silver before:", before_count)
print("Prepared records:", prepared.count())
print("Silver after:", after_count)
print("New records inserted:", after_count - before_count)
```

```python

from pyspark.sql import functions as F
from pyspark.sql.window import Window

# ============================================================
# 1. LOAD SILVER TABLES
# ============================================================

pollution = spark.table("workspace.smartair.silver_openaq_measurements")
weather = spark.table("workspace.smartair.silver_weather")

print("Silver Pollution Records:", pollution.count())
print("Silver Weather Records:", weather.count())


# ============================================================
# 2. GOLD POLLUTANT ANALYSIS
# ============================================================

gold_pollutant = (
    pollution
    .groupBy("pollutant", "unit")
    .agg(
        F.count("*").alias("total_measurements"),
        F.avg("concentration").alias("average_concentration"),
        F.min("concentration").alias("minimum_concentration"),
        F.max("concentration").alias("maximum_concentration"),
        F.countDistinct("sensor_id").alias("unique_sensors")
    )
    .withColumn("gold_created_at", F.current_timestamp())
    .select(
        "pollutant",
        "unit",
        "total_measurements",
        "average_concentration",
        "minimum_concentration",
        "maximum_concentration",
        "unique_sensors",
        "gold_created_at"
    )
)


# ============================================================
# 3. GOLD DAILY POLLUTION TRENDS
# ============================================================

gold_daily = (
    pollution
    .withColumn(
        "measurement_date",
        F.to_date("measurement_start_utc")
    )
    .groupBy(
        "search_center_city",
        "actual_location",
        "pollutant",
        "unit",
        "measurement_date"
    )
    .agg(
        F.count("*").alias("total_measurements"),
        F.avg("concentration").alias("average_concentration"),
        F.min("concentration").alias("minimum_concentration"),
        F.max("concentration").alias("maximum_concentration")
    )
    .withColumn("gold_created_at", F.current_timestamp())
    .select(
        "search_center_city",
        "actual_location",
        "pollutant",
        "unit",
        "measurement_date",
        "total_measurements",
        "average_concentration",
        "minimum_concentration",
        "maximum_concentration",
        "gold_created_at"
    )
)


# ============================================================
# 4. GOLD POLLUTION HOTSPOTS
# Top 10 locations for each pollutant
# ============================================================

hotspot_window = (
    Window
    .partitionBy("pollutant")
    .orderBy(F.desc("average_concentration"))
)

gold_hotspots = (
    pollution
    .groupBy("actual_location", "pollutant", "unit")
    .agg(
        F.avg("concentration").alias("average_concentration"),
        F.count("*").alias("total_measurements")
    )
    .withColumn(
        "pollution_rank",
        F.row_number().over(hotspot_window).cast("int")
    )
    .filter(F.col("pollution_rank") <= 10)
    .withColumn("gold_created_at", F.current_timestamp())
    .select(
        "actual_location",
        "pollutant",
        "unit",
        "average_concentration",
        "total_measurements",
        "pollution_rank",
        "gold_created_at"
    )
)


# ============================================================
# 5. GOLD CITY SUMMARY
# ============================================================

gold_city = (
    pollution
    .groupBy("search_center_city", "pollutant", "unit")
    .agg(
        F.count("*").alias("total_measurements"),
        F.countDistinct("actual_location").alias("total_locations"),
        F.countDistinct("sensor_id").alias("total_sensors"),
        F.avg("concentration").alias("average_concentration"),
        F.min("concentration").alias("minimum_concentration"),
        F.max("concentration").alias("maximum_concentration")
    )
    .withColumn("gold_created_at", F.current_timestamp())
    .select(
        "search_center_city",
        "pollutant",
        "unit",
        "total_measurements",
        "total_locations",
        "total_sensors",
        "average_concentration",
        "minimum_concentration",
        "maximum_concentration",
        "gold_created_at"
    )
)


# ============================================================
# 6. GOLD WEATHER-POLLUTION ANALYSIS
# Join pollution and weather using city + measurement hour
# ============================================================

pollution_hourly = (
    pollution
    .withColumn(
        "measurement_hour",
        F.date_trunc("hour", F.col("measurement_start_utc"))
    )
)

weather_hourly = (
    weather
    .withColumn(
        "weather_hour",
        F.date_trunc("hour", F.col("measurement_time"))
    )
    .groupBy("city", "weather_hour")
    .agg(
        F.avg("temperature_c").alias("temperature_c"),
        F.avg("relative_humidity_percent").alias("relative_humidity_percent"),
        F.avg("wind_speed_kmh").alias("wind_speed_kmh"),
        F.avg("precipitation_mm").alias("precipitation_mm"),
        F.avg("rain_mm").alias("rain_mm")
    )
)

p = pollution_hourly.alias("p")
w = weather_hourly.alias("w")

weather_joined = (
    p.join(
        w,
        (
            F.lower(F.trim(F.col("p.search_center_city"))) ==
            F.lower(F.trim(F.col("w.city")))
        ) &
        (
            F.col("p.measurement_hour") ==
            F.col("w.weather_hour")
        ),
        "inner"
    )
)

gold_weather = (
    weather_joined
    .groupBy(
        F.col("p.measurement_hour").alias("measurement_hour"),
        F.col("p.actual_location").alias("actual_location"),
        F.col("p.pollutant").alias("pollutant"),
        F.col("p.unit").alias("unit"),
        F.col("w.temperature_c").alias("temperature_c"),
        F.col("w.relative_humidity_percent").alias("relative_humidity_percent"),
        F.col("w.wind_speed_kmh").alias("wind_speed_kmh"),
        F.col("w.precipitation_mm").alias("precipitation_mm"),
        F.col("w.rain_mm").alias("rain_mm")
    )
    .agg(
        F.avg("p.concentration").alias("hourly_average_concentration"),
        F.count("*").alias("pollution_readings")
    )
    .withColumn("gold_created_at", F.current_timestamp())
    .select(
        "measurement_hour",
        "actual_location",
        "pollutant",
        "unit",
        "hourly_average_concentration",
        "pollution_readings",
        "temperature_c",
        "relative_humidity_percent",
        "wind_speed_kmh",
        "precipitation_mm",
        "rain_mm",
        "gold_created_at"
    )
)


# ============================================================
# 7. GOLD POLLUTION ALERTS
# Preserve existing PM10 and PM2.5 thresholds
# ============================================================

gold_alerts = (
    pollution
    .filter(F.col("pollutant").isin("pm10", "pm25"))
    .withColumn(
        "alert_level",
        F.when(
            (F.col("pollutant") == "pm10") &
            (F.col("concentration") < 50),
            "Low"
        )
        .when(
            (F.col("pollutant") == "pm10") &
            (F.col("concentration") < 100),
            "Moderate"
        )
        .when(
            (F.col("pollutant") == "pm10"),
            "High"
        )
        .when(
            (F.col("pollutant") == "pm25") &
            (F.col("concentration") <= 30),
            "Low"
        )
        .when(
            (F.col("pollutant") == "pm25") &
            (F.col("concentration") <= 60),
            "Moderate"
        )
        .otherwise("High")
    )
    .withColumn("alert_created_at", F.current_timestamp())
    .select(
        "sensor_id",
        "actual_location",
        "search_center_city",
        "pollutant",
        "concentration",
        "unit",
        "measurement_start_utc",
        "measurement_end_utc",
        "data_completeness",
        "observed_count",
        "silver_ingested_at",
        "alert_level",
        "alert_created_at"
    )
)


# ============================================================
# 8. REFRESH ALL SIX GOLD TABLES
# Existing table names are retained
# ============================================================

gold_tables = [
    (gold_pollutant, "gold_pollutant_analysis"),
    (gold_daily, "gold_daily_aqi_trends"),
    (gold_hotspots, "gold_pollution_hotspots"),
    (gold_city, "gold_city_aqi_summary"),
    (gold_weather, "gold_weather_pollution_analysis"),
    (gold_alerts, "gold_aqi_alerts")
]

for dataframe, table_name in gold_tables:
    full_table_name = f"workspace.smartair.{table_name}"

    (
        dataframe.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable(full_table_name)
    )

    print(f"REFRESHED: {full_table_name}")


print("\n========== GOLD LAYER REFRESH COMPLETED ==========")
```
